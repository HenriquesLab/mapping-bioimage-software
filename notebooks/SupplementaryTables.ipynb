## Set up offline table generation

Locate the standalone repository or the manuscript scripts checkout. Import the publication table builders and read the SI once if it exists. No default cell writes data or calls a collector.

```python
from pathlib import Path
import re
import sys
from IPython.display import Markdown, display

cwd = Path.cwd().resolve()
SCRIPTS_DIR = next(p for p in (cwd, *cwd.parents) if (p / 'census' / '_registry.py').is_file())
if str(SCRIPTS_DIR) not in sys.path:
    sys.path.insert(0, str(SCRIPTS_DIR))
from census.tables import make_si_tables as tables

si_path = SCRIPTS_DIR.parent / 'MANUSCRIPT' / '02_SUPPLEMENTARY_INFO.md'
si_text = si_path.read_text() if si_path.is_file() else None
rendered_tables = {}
checked_tables = set()

def check_table(name, body):
    assert body.startswith('| ') and len(body.splitlines()) > 2, name
    if si_text is None:
        print(f'{name}: manuscript comparison skipped; SI is absent in this clone')
        return
    match = re.search(rf'<!-- TABLE:{name} START -->\s*\n(.*?)\n<!-- TABLE:{name} END -->', si_text, re.S)
    assert match is not None, f'{name}: manuscript table markers missing'
    assert body.strip() == match.group(1).strip(), f'{name}: differs from manuscript table'
    checked_tables.add(name)
    print(f'{name}: identical to manuscript table')
```

## Rebuilding from scratch

The default cells read committed inputs and run offline. The producer commands in Inputs rebuild those files. Source-count collectors and new annotation runs need network access; rebuilding sampled article text first requires `python release/refetch_identifiers.py`. Classification, validation pooling and table generation can then run locally from the recovered corpus and saved labels. Preserve the original seeds, sample frames and collection dates when checking the published snapshot.

`python -m census.validation.replicates pool` pools saved validation labels. `python -m census.tables.make_si_tables` writes the resulting tables into an existing manuscript SI file; this notebook only displays and checks them.
